```python
# Real-world example: Movie recommendation confidence.
# 'user_sci_fi_interest' is the weight (how much we think the user likes the genre),
# 'movie_sci_fi_score' is the input (how much the movie belongs to that genre),
# and 'actual_user_rating_for_this_movie' is the user's actual rating of the movie.

movie_sci_fi_score = 0.8    # Movie is 80% Sci-Fi
user_sci_fi_interest = 0.3  # How much we think the User loves Sci-Fi (normalized)
actual_user_rating_for_this_movie = 0.85 # User actually rated it 8/10

prediction = user_sci_fi_interest * movie_sci_fi_score
error = (prediction - actual_user_rating_for_this_movie) ** 2

print(f"Predicted User Rating for sci-fi movie: {prediction}")
print(f"Actual User Rating for this movie: {actual_user_rating_for_this_movie}")
print(f"Error (Squared Difference): {error:.10f}")
```

```python
# Hot and Cold Learning - means wiggling the weights to see which direction reduces the error the most,
# then moving the weights in that direction and repeating until the error gets to 0.

input = 0.5
weight = 0.5
target = 0.8
learning_step_amount = 0.15

def neural_network(input, weight):
    prediction = input * weight
    error = (prediction - target) ** 2 # Forces the error to be positive by multiplying it by itself. Negative error wouldn't make sense.
    # print(f"Predicted value: {prediction:.5f}")
    # print(f"Actual value: {target:.5f}")
    # print(f"Error: {error:.5f}")
    return error

def hot_and_cold(input, weight, learning_step_amount):
    for i in range(20):
        down_weight = weight - learning_step_amount
        down_error = neural_network(input, down_weight)
        print(f"  down_error: {down_error:.5f}")
        up_weight = weight + learning_step_amount
        up_error = neural_network(input, up_weight)
        print(f"  up_error: {up_error:.5f}")
        if down_error < up_error:
            print(f"---  Decreasing weight -{learning_step_amount}")
            weight -= learning_step_amount
        else:
            print(f"---  Increasing weight +{learning_step_amount}")
            weight += learning_step_amount
    return weight

initial_error = neural_network(input, weight)
print(f"=======\nInitial Error: {initial_error:.5f}")
print(f"Initial Weight: {weight:.5f}\n=======")

print("Starting Hot and Cold training...")
weight = hot_and_cold(input, weight, learning_step_amount)

after_training_error = neural_network(input, weight)
print(f"=======\nAfter Training Error: {after_training_error:.5f}")
print(f"Final Weight: {weight:.5f}\n=======")
```

```python
# Gradient Descent learning - better than Hot and Cold because of fewer iterations and will never overshoot and more precise error decrease

input = 0.5
weight = 0.5
target = 0.8


def neural_network(input, weight):
    prediction = input * weight
    error = (prediction - target) ** 2
    direction_and_amount = (prediction - target) * input # Scaling, negative reversal, and stopping
    print(f"    Predicted value: {prediction:.5f}")
    print(f"    Actual value: {target:.5f}")
    print(f"    Error: {error:.5f}")
    print(f"    Direction and amount: {direction_and_amount:.5f}")
    return direction_and_amount

def gradient_descent(input, weight):
    for i in range(20):
        direction_and_amount = neural_network(input, weight)
        weight -= direction_and_amount
        print(f"- changing weight by: {direction_and_amount:.5f}")
    return weight

initial_error = neural_network(input, weight)
print(f"=======\nInitial Error: {initial_error:.5f}")
print(f"Initial Weight: {weight:.5f}\n=======")

print("Starting Gradient Descent training...")
weight = gradient_descent(input, weight)

after_training_error = neural_network(input, weight)
print(f"=======\nAfter Training Error: {after_training_error:.5f}")
print(f"Final Weight: {weight:.5f}\n=======")
```

```python
# Introducing Alpha - to be able to control the speed of training to prevent overshooting if weight_delta is higher than delta because of a very big input value

# Let's rewrite the code above to have all the official deep learning nomenclature

input = 2
weight = 0.5
target = 0.8
alpha = 0.1

for iteration in range(10):
    prediction = input * weight
    error = (prediction - target) ** 2
    delta = prediction - target
    weight_delta = delta * input
    weight = weight - (alpha * weight_delta)
    print(f"    Predicted value: {prediction:.5f}")
    print(f"    Actual value: {target:.5f}")
    print(f"    Error: {error:.10f}")
    print(f"    Direction and amount: {weight_delta:.5f}\n----")
```
