## How Neural Networks PREDICT

In this chapter we will explore the concept of **dot product** and how it can be used to calculate **similarity** between vectors.
We will also learn how to use **NumPy library** to perform vector operations efficiently.
And we will finish with a real-life example: **Movie Recommendation Engine.**

```python
weight = 0.1

def neural_network(input, weight):
    prediction = input * weight
    return prediction

number_of_toes = [8.5,9.5,10,9]
input = number_of_toes[0]
pred = neural_network(input, weight)
print(pred)
```

```python
# Dot Product - Is vector_a a weighted sum between two vectors

weights = [0.1,0.2,0]

def weighted_sum(vector_a,vector_b):
    assert(len(vector_a) == len(vector_b))
    output = 0
    for i in range (len(vector_a)):
        output += vector_a[i] * vector_b[i]

    return output

def neural_network(inputs, weights):
    pred = weighted_sum(inputs,weights)
    return pred

number_of_toes = [8.5,9.5,10,9]
wlrec = [0.65,0.8,0.8,0.9]
nfans = [1.2,1.3,0.5,1.0]

inputs = [number_of_toes[0],wlrec[0],nfans[0]]
pred = neural_network(inputs, weights)
print(pred)
```

```python
# Dot Product gives you vector_a `notion of similarity`

vector_a = [0,1,0,1]
vector_b = [1,0,1,0]
c = [0,1,1,0]
d = [0.5,0,0.5,0]
e = [0,1,-1,0]

print(weighted_sum(vector_a,vector_b))
print(weighted_sum(vector_b,c))
print(weighted_sum(vector_b,d))
print(weighted_sum(c,c))
print(weighted_sum(d,d))
print(weighted_sum(c,e))
```

```python
# Above example but using NumPy library

import numpy as np

weights = np.array([0.1,0.2,0])
def neural_network(inputs,weights):
    pred = inputs.dot(weights)
    return pred

number_of_toes = np.array([8.5,9.5,10,9])
wlrec = np.array([0.65,0.8,0.8,0.9])
nfans = np.array([1.2,1.3,0.5,1.0])

inputs = np.array([number_of_toes[0],wlrec[0],nfans[0]])
pred = neural_network(inputs,weights)
print(pred)
```

```python
# single input, multiple outputs

weights = [0.1,0.2,0]

def elementwise_multiplication(number,vector):
    output = [0,0,0]
    assert(len(output) == len(vector))
    for i in range (len(vector)):
        output[i] = number * vector[i]

    return output

def neural_network(number, weights):
    pred = elementwise_multiplication(number,weights)
    return pred

input = 0.65
pred = neural_network(input, weights)
print(pred)
```

```python
# Vector Matrix Multiplication - multiple input, multiple output

weights = [
    [0.1,0.1,-0.3],
    [0.1,0.2,0.0],
    [0.1,0.2,0.0]]

def dot_product(vector_a, vector_b):
    assert(len(vector_a) == len(vector_b))
    output = 0
    for i in range (len(vector_a)):
        output += vector_a[i] * vector_b[i]
    return output

def vector_matrix_multiplication(vector, matrix):
    assert(len(vector) == len(matrix[0]))
    output = [0,0,0]
    for i in range (len(vector)):
        output[i] = dot_product(vector,matrix[i])
    return output

def neural_network(inputs, weights):
    pred = vector_matrix_multiplication(inputs,weights)
    return pred

number_of_toes = [8.5,9.5,10,9]
wlrec = [0.65,0.8,0.8,0.9]
nfans = [1.2,1.3,0.5,1.0]

inputs = [number_of_toes[0],wlrec[0],nfans[0]]

pred = neural_network(inputs, weights)
print(pred)

```

```python
# Predicting on predictions - neural networks can be stacked!

inputs_hiddens_weights = [
    [ 0.1, 0.2,-0.1],
    [-0.1, 0.1, 0.9],
    [ 0.1, 0.4, 0.1]
]

hiddens_predictions_weights = [
    [0.3, 1.1,-0.3],
    [0.1, 0.2, 0.0],
    [0.0, 1.3, 0.1]
]

weights = [inputs_hiddens_weights, hiddens_predictions_weights]

def dot_product(vector_a, vector_b):
    assert(len(vector_a) == len(vector_b))
    output = 0
    for i in range (len(vector_a)):
        output += vector_a[i] * vector_b[i]
    return output

def vector_matrix_multiplication(vector, matrix):
    assert(len(vector) == len(matrix))
    output = [0,0,0]
    for i in range (len(vector)):
        output[i] = dot_product(vector,matrix[i])
    return output

def neural_network(input, weights):
    hidden = vector_matrix_multiplication(input, weights[0])
    predictions = vector_matrix_multiplication(hidden, weights[1]) # <-- hidden is input for the second layer
    return predictions

number_of_toes = [8.5,9.5,10,9]
wlrec = [0.65,0.8,0.8,0.9]
nfans = [1.2,1.3,0.5,1.0]

input = [number_of_toes[0],wlrec[0],nfans[0]]
pred = neural_network(input, weights)
print(pred)
```

```python
# Same example as above but - using numpy

import numpy as np
inputs_hiddens_weights = np.array([
    [ 0.1, 0.2,-0.1],
    [-0.1, 0.1, 0.9],
    [ 0.1, 0.4, 0.1]
]).T

hiddens_predictions_weights = np.array([
    [0.3, 1.1,-0.3],
    [0.1, 0.2, 0.0],
    [0.0, 1.3, 0.1]
]).T

weights = [inputs_hiddens_weights, hiddens_predictions_weights]

def neural_network(input, weights):
    hidden = input.dot(weights[0])
    predictions = hidden.dot(weights[1])
    return predictions

number_of_toes = np.array([8.5,9.5,10,9])
wlrec = np.array([0.65,0.8,0.8,0.9])
nfans = np.array([1.2,1.3,0.5,1.0])

input = np.array([number_of_toes[0],wlrec[0],nfans[0]])
pred = neural_network(input, weights)
print(pred)
```

### Real-World Example: Movie Recommendation Engine
Imagine we want to predict a user's interest score for library of movies.
- **Inputs**: The movie's profile (how much Comedy, Horror, Sci-Fi, Romance it contains, from 0 to 1).
- **Weights**: The user's personal preference for each genre. (how much Comedy, Horror, Sci-Fi, Romance he likes, from 0 to 1).
- **Output**: Interest Score

```python
# Movie Profiles
#                          [Comedy, Horror, Sci-Fi, Romance]
scary_movie =     np.array([0.8,    0.2,    0.1,    0.4])  # High comedy, low horror, little romance
deadspace =       np.array([0.0,    1.0,    0.9,    0.0])  # High horror high sci-fi
lame_chickflick = np.array([0.4,    0.0,    0.0,    0.9])  # High Romance, some comedy

# User Preferences: Likes Sci-Fi and Horror, dislikes Romance
#                     [Comedy, Horror, Sci-Fi, Romance]
user_profile = np.array([0.4,    0.9,    0.8,    -0.5])


def calculate_recommendation_score(movie_name, movie_profile, user_profile):
    score = movie_profile.dot(user_profile)
    print(f"{movie_name}: Score = {score:.2f}")
    if score > 0.8:
        return "Must Watch!"
    elif score > 0.3:
        return "Worth a try."
    else:
        return "Not for you."


print(f"{calculate_recommendation_score('Scary Movie', scary_movie, user_profile)}\n")
print(f"{calculate_recommendation_score('Deadspace', deadspace, user_profile)}\n")
print(f"{calculate_recommendation_score('Some Chickflick', lame_chickflick, user_profile)}\n")
```

The intuition behind how and why a **dot product** works is that it gives you a measure of **similarity** between two vectors (in this case, vector1: the movie profile and vector2: the user profile).
